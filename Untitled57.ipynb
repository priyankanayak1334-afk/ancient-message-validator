# Ancient Message Validator 📜

## 📌 Project Overview
This project simulates decoding an ancient civilization's messaging system that relies on nested brackets. It utilizes a **Stack (LIFO)** data structure to validate whether strings of brackets (`()`, `{}`, `[]`) are properly closed and nested.

## ⚙️ How the Stack Behaves
1. **Push:** When an opening bracket (`(`, `{`, `[`) is encountered, it is pushed onto the stack.
2. **Peek & Pop:** When a closing bracket appears, the code checks the top element of the stack. If it matches, the opening bracket is popped off.
3. **Early Failure:** If a closing bracket appears and the stack is empty, or doesn't match the top element, it is immediately flagged as **invalid**.
4. **Final Check:** After parsing the entire string, the stack must be completely **empty** to be considered valid.


```python
def is_balanced(message: str) -> bool:
    # Map closing brackets to their corresponding opening brackets
    bracket_map = {')': '(', '}': '{', ']': '['}
    stack = []

    for char in message:
        # If it's an opening bracket, push it onto the stack
        if char in bracket_map.values():
            stack.append(char)
        # If it's a closing bracket
        elif char in bracket_map:
            # Check if stack is empty or top of stack doesn't match
            if not stack or stack[-1] != bracket_map[char]:
                return False
            stack.pop() # Successfully matched, remove from stack
            
    # The sequence is valid only if the stack is completely empty
    return not stack
```

```python
# Define test edge cases
test_cases = {
    "{}()[]": True,       # Standard balanced
    "{[()]}": True,       # Nested balanced
    "{[(])}": False,      # Incorrect nesting order
    "{[(": False,         # Unclosed brackets
    "()]": False,         # Extra closing brackets
    "Ancient[text]": True # Ignores non-bracket characters
}

print("--- Running Bracket Validation Tests ---")
all_passed = True
for i, (text, expected) in enumerate(test_cases.items(), 1):
    result = is_balanced(text)
    passed = result == expected
    print(f"Test {i}: '{text}' -> Passed: {passed} (Got: {result})")
    if not passed:
        all_passed = False
        
print(f"\nResult: {'🎉 ALL TESTS PASSED' if all_passed else '❌ SOME TESTS FAILED'}")
```
